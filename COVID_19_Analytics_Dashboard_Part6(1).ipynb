# COVID-19 Analytics Dashboard
## Part 6 - Exploratory Data Analysis (EDA) with Matplotlib

This notebook section visualizes the COVID-19 data using Matplotlib.


```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('country_cleaned.csv')
print(df.shape)
display(df.head())
```

```python
if 'Confirmed' in df.columns:
    top10 = df.sort_values('Confirmed', ascending=False).head(10)
    plt.figure(figsize=(10,5))
    plt.bar(top10.iloc[:,0], top10['Confirmed'])
    plt.xticks(rotation=90)
    plt.title('Top 10 Countries by Confirmed Cases')
    plt.xlabel('Country')
    plt.ylabel('Confirmed Cases')
    plt.tight_layout()
    plt.show()
```

```python
for col in ['Confirmed','Deaths','Recovered','Active']:
    if col in df.columns:
        plt.figure(figsize=(7,4))
        plt.hist(df[col], bins=20)
        plt.title(f'{col} Distribution')
        plt.xlabel(col)
        plt.ylabel('Frequency')
        plt.show()
```

```python
if {'Confirmed','Deaths'}.issubset(df.columns):
    plt.figure(figsize=(6,5))
    plt.scatter(df['Confirmed'], df['Deaths'])
    plt.title('Confirmed vs Deaths')
    plt.xlabel('Confirmed')
    plt.ylabel('Deaths')
    plt.show()

if {'Confirmed','Recovered'}.issubset(df.columns):
    plt.figure(figsize=(6,5))
    plt.scatter(df['Confirmed'], df['Recovered'])
    plt.title('Confirmed vs Recovered')
    plt.xlabel('Confirmed')
    plt.ylabel('Recovered')
    plt.show()
```

```python
cols=[c for c in ['Confirmed','Deaths','Recovered','Active'] if c in df.columns]
if cols:
    totals=df[cols].sum()
    plt.figure(figsize=(6,6))
    plt.pie(totals.values,labels=totals.index,autopct='%1.1f%%')
    plt.title('Overall COVID-19 Distribution')
    plt.show()
```
