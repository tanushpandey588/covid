```python
import pandas as pd
import matplotlib.pyplot as plt

country=pd.read_csv('country_cleaned.csv')
world=pd.read_csv('world_cleaned.csv')
day=pd.read_csv('day_cleaned.csv')
print('Datasets loaded successfully.')
```

```python
print('='*60)
print('FINAL PROJECT SUMMARY')
print('='*60)
print('Country Dataset Shape :',country.shape)
print('Worldometer Dataset Shape :',world.shape)
print('Day-wise Dataset Shape :',day.shape)

if 'Confirmed' in country.columns:
    print('Total Confirmed Cases :',country['Confirmed'].sum())
if 'Deaths' in country.columns:
    print('Total Deaths :',country['Deaths'].sum())
if 'Recovered' in country.columns:
    print('Total Recoveries :',country['Recovered'].sum())
```

```python
numeric=country.select_dtypes(include='number')
if not numeric.empty:
    avg=numeric.mean()
    print('Average Values')
    print(avg)
```

```python
if {'Confirmed','Deaths','Recovered'}.issubset(country.columns):
    totals=[country['Confirmed'].sum(),country['Deaths'].sum(),country['Recovered'].sum()]
    labels=['Confirmed','Deaths','Recovered']
    plt.figure(figsize=(7,5))
    plt.bar(labels,totals)
    plt.title('Overall COVID-19 Summary')
    plt.xlabel('Category')
    plt.ylabel('Cases')
    plt.show()
```
