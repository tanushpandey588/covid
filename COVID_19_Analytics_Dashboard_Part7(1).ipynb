# COVID-19 Analytics Dashboard
## Part 7 - Advanced Visualizations & Time-Series Analysis


```python
import pandas as pd
import matplotlib.pyplot as plt

day_df=pd.read_csv('day_cleaned.csv')
country_df=pd.read_csv('country_cleaned.csv')
print(day_df.head())
```

```python
if 'Date' in day_df.columns:
    day_df['Date']=pd.to_datetime(day_df['Date'])
    day_df=day_df.sort_values('Date')
    for col in ['Confirmed','Deaths','Recovered','Active']:
        if col in day_df.columns:
            plt.figure(figsize=(10,4))
            plt.plot(day_df['Date'],day_df[col])
            plt.title(f'{col} Over Time')
            plt.xticks(rotation=45)
            plt.tight_layout()
            plt.show()
```

```python
if {'Confirmed','Deaths'}.issubset(country_df.columns):
    top=country_df.nlargest(15,'Confirmed')
    plt.figure(figsize=(11,5))
    plt.bar(top.iloc[:,0],top['Confirmed'])
    plt.xticks(rotation=90)
    plt.title('Top 15 Countries by Confirmed Cases')
    plt.tight_layout();plt.show()
    plt.figure(figsize=(11,5))
    plt.bar(top.iloc[:,0],top['Deaths'])
    plt.xticks(rotation=90)
    plt.title('Deaths in Top 15 Countries')
    plt.tight_layout();plt.show()
```

```python
if {'Confirmed','Recovered','Deaths'}.issubset(country_df.columns):
    country_df['Recovery Rate (%)']=country_df['Recovered']/country_df['Confirmed']*100
    country_df['Death Rate (%)']=country_df['Deaths']/country_df['Confirmed']*100
    top_rr=country_df.nlargest(10,'Recovery Rate (%)')
    plt.figure(figsize=(10,5))
    plt.bar(top_rr.iloc[:,0],top_rr['Recovery Rate (%)'])
    plt.xticks(rotation=90)
    plt.title('Top Recovery Rate')
    plt.tight_layout();plt.show()
```
