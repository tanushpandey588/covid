# COVID-19 Analytics Dashboard
## Part 4 - Matplotlib Visualizations

```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

country_df = pd.read_csv("country_wise_latest.csv")
```

```python
## Line Plot
```

```python
plt.figure(figsize=(10,5))
plt.plot(country_df["Confirmed"])
plt.title("Confirmed Cases")
plt.xlabel("Countries")
plt.ylabel("Confirmed")
plt.grid(True)
plt.show()
```

## Bar Chart

```python
top10 = country_df.nlargest(10, "Confirmed")
plt.figure(figsize=(10,5))
plt.bar(top10["Country/Region"], top10["Confirmed"])
plt.xticks(rotation=45)
plt.title("Top 10 Countries by Confirmed Cases")
plt.tight_layout()
plt.show()
```

## Pie Chart

```python
top5 = country_df.nlargest(5, "Deaths")
plt.figure(figsize=(7,7))
plt.pie(top5["Deaths"], labels=top5["Country/Region"], autopct="%1.1f%%")
plt.title("Top 5 Countries by Deaths")
plt.show()
```

## Histogram

```python
plt.figure(figsize=(8,5))
plt.hist(country_df["Confirmed"], bins=20)
plt.title("Distribution of Confirmed Cases")
plt.show()
```

## Scatter Plot

```python
plt.figure(figsize=(8,5))
plt.scatter(country_df["Confirmed"], country_df["Deaths"])
plt.xlabel("Confirmed")
plt.ylabel("Deaths")
plt.title("Confirmed vs Deaths")
plt.show()
```

## Box Plot

```python
plt.figure(figsize=(6,5))
plt.boxplot(country_df["Confirmed"])
plt.title("Confirmed Cases Box Plot")
plt.show()
```

## Multiple Line Plot

```python
plt.figure(figsize=(10,5))
plt.plot(country_df["Confirmed"], label="Confirmed")
plt.plot(country_df["Recovered"], label="Recovered")
plt.plot(country_df["Deaths"], label="Deaths")
plt.legend()
plt.grid(True)
plt.title("COVID-19 Comparison")
plt.show()
```

## Save Figure

```python
plt.figure(figsize=(8,4))
plt.plot(country_df["Confirmed"])
plt.title("Confirmed Cases")
plt.savefig("confirmed_cases.png")
plt.close()
print("Graph saved as confirmed_cases.png")
```
