# COVID-19 Analytics Dashboard
## Part 8 - Worldometer Analysis & Country-wise Statistical Insights

This part focuses on analyzing the **worldometer_data** dataset and generating country-wise rankings and insights.


```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

world = pd.read_csv('world_cleaned.csv')
print('Dataset Shape:', world.shape)
display(world.head())
```

```python
print('Columns:')
print(world.columns.tolist())

print('\nSummary Statistics')
display(world.describe(include='all'))

print('\nMissing Values')
print(world.isnull().sum())
```

```python
for col in ['TotalCases','TotalDeaths','TotalRecovered']:
    if col in world.columns:
        top = world.sort_values(col, ascending=False).head(10)
        print(f'Top 10 by {col}')
        display(top)
        plt.figure(figsize=(10,5))
        plt.bar(top.iloc[:,0], top[col])
        plt.xticks(rotation=90)
        plt.title(f'Top 10 Countries by {col}')
        plt.tight_layout()
        plt.show()
```

```python
numeric = world.select_dtypes(include='number')
print('Correlation Matrix')
display(numeric.corr())

print('Covariance Matrix')
display(numeric.cov())
```

```python
if {'TotalCases','TotalDeaths'}.issubset(world.columns):
    plt.figure(figsize=(7,5))
    plt.scatter(world['TotalCases'], world['TotalDeaths'])
    plt.title('Total Cases vs Total Deaths')
    plt.xlabel('Total Cases')
    plt.ylabel('Total Deaths')
    plt.show()
```
