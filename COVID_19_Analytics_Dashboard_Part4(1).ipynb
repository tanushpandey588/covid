# COVID-19 Analytics Dashboard
## Part 4 - NumPy Analysis

This section demonstrates how NumPy can be used for numerical analysis on COVID-19 datasets.


```python
import numpy as np
import pandas as pd

# Load cleaned dataset (generated in Part 3)
df = pd.read_csv('country_cleaned.csv')
print('Dataset Shape:', df.shape)
display(df.head())
```

```python
# Select numeric columns
numeric_df = df.select_dtypes(include='number')
arr = numeric_df.to_numpy()
print('NumPy Array Shape:', arr.shape)
print(arr[:5])

print('\nBasic Statistics')
print('Minimum:', np.min(arr, axis=0))
print('Maximum:', np.max(arr, axis=0))
print('Mean:', np.mean(arr, axis=0))
print('Median:', np.median(arr, axis=0))
print('Standard Deviation:', np.std(arr, axis=0))
print('Variance:', np.var(arr, axis=0))

print('\nPercentiles')
print('25%:', np.percentile(arr,25,axis=0))
print('50%:', np.percentile(arr,50,axis=0))
print('75%:', np.percentile(arr,75,axis=0))

print('\nRow-wise Sum (First 10):')
print(np.sum(arr, axis=1)[:10])

print('\nColumn-wise Sum:')
print(np.sum(arr, axis=0))

print('\nCorrelation Matrix')
print(np.corrcoef(arr, rowvar=False))
```
