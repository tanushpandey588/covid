# COVID-19 Analytics Dashboard
## Part 2 - Loading & Exploring Datasets


```python
import os
import pandas as pd

files={
'country':'country_wise_latest.csv',
'clean':'covid_19_clean_complete.csv',
'day':'day_wise.csv',
'world':'worldometer_data.csv'}
for k,v in files.items():
    print(f'{k}:',v,'Exists' if os.path.exists(v) else 'Not Found')
```

```python
datasets={}
for k,v in files.items():
    try:
        datasets[k]=pd.read_csv(v)
        print(f'Loaded {v}')
    except Exception as e:
        print(f'Could not load {v}: {e}')
```

```python
for name,df in datasets.items():
    print('\n'+'='*60)
    print(name.upper())
    print('='*60)
    print('Shape:',df.shape)
    print('\nHead:')
    display(df.head())
    print('\nTail:')
    display(df.tail())
    print('\nColumns:')
    print(df.columns.tolist())
    print('\nInfo:')
    df.info()
    print('\nDescribe:')
    display(df.describe(include='all'))
    print('\nMissing Values:')
    print(df.isnull().sum())
    print('\nDuplicate Rows:',df.duplicated().sum())
    print('\nData Types:')
    print(df.dtypes)
    print('\nMemory Usage (bytes):')
    print(df.memory_usage(deep=True).sum())
```
