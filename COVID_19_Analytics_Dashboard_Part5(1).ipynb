# COVID-19 Analytics Dashboard
## Part 5 - Advanced Pandas Analysis

In this part we perform data manipulation and exploratory analysis using Pandas.


```python
import pandas as pd
import numpy as np

df = pd.read_csv('country_cleaned.csv')
print('Shape:', df.shape)
display(df.head())
```

```python
print('Columns:')
print(df.columns.tolist())

print('\nData Types')
print(df.dtypes)

print('\nStatistical Summary')
display(df.describe(include='all'))

print('\nMissing Values')
print(df.isnull().sum())
```

```python
numeric = df.select_dtypes(include='number')
print('Mean Values')
print(numeric.mean())

print('\nMedian Values')
print(numeric.median())

print('\nMaximum Values')
print(numeric.max())

print('\nMinimum Values')
print(numeric.min())
```

```python
if 'Confirmed' in df.columns:
    top_confirmed = df.sort_values('Confirmed', ascending=False).head(10)
    print('Top 10 Countries by Confirmed Cases')
    display(top_confirmed)

if 'Deaths' in df.columns:
    top_deaths = df.sort_values('Deaths', ascending=False).head(10)
    print('Top 10 Countries by Deaths')
    display(top_deaths)

if 'Recovered' in df.columns:
    top_recovered = df.sort_values('Recovered', ascending=False).head(10)
    print('Top 10 Countries by Recoveries')
    display(top_recovered)
```

```python
print('Correlation Matrix')
display(numeric.corr())

print('Covariance Matrix')
display(numeric.cov())
```

- Calculated summary statistics.
- Identified top countries by confirmed cases, deaths, and recoveries.
