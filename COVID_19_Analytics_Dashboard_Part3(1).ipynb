# COVID-19 Analytics Dashboard
## Part 3 - Data Cleaning & Preprocessing


```python
import pandas as pd
import numpy as np

files=['country_wise_latest.csv','covid_19_clean_complete.csv','day_wise.csv','worldometer_data.csv']
dfs=[pd.read_csv(f) for f in files]
names=['country','clean','day','world']
for name,df in zip(names,dfs):
    print('\nCleaning',name)
    print('Original Shape:',df.shape)
    df.columns=df.columns.str.strip()
    obj=df.select_dtypes(include='object').columns
    for c in obj:
        df[c]=df[c].astype(str).str.strip()
    print('Missing Values Before:')
    print(df.isnull().sum())
    for c in df.columns:
        if df[c].dtype=='object':
            df[c]=df[c].fillna('Unknown')
        else:
            df[c]=df[c].fillna(df[c].median())
    print('Missing Values After:')
    print(df.isnull().sum())
    dup=df.duplicated().sum()
    print('Duplicates:',dup)
    if dup>0:
        df.drop_duplicates(inplace=True)
    print('New Shape:',df.shape)
    print('Data Types:')
    print(df.dtypes)
    print('Summary:')
    display(df.describe(include='all'))
    out=name+'_cleaned.csv'
    df.to_csv(out,index=False)
    print('Saved:',out)
```

All datasets cleaned and exported for further analysis.
